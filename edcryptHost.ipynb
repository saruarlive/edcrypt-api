## Just for self testing, Compare Hashes (for Encrypted/Decrypted Verification)

```python
import hashlib

def sha256sum(filename):
    h = hashlib.sha256()
    with open(filename, "rb") as f:
        for chunk in iter(lambda: f.read(8192), b""):
            h.update(chunk)
    return h.hexdigest()

h1 = sha256sum("naif-api-details.txt")
h2 = sha256sum("decrypt-naif-key.txt")

# print("Original :", h1)
# print("Decrypted:", h2)

print("✅ Match" if h1 == h2 else "❌ Different")
```
